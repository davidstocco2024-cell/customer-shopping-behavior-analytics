```python
labels= ['Young Adult', 'Adult', 'Middle_aged', 'Senior']
df['age_group'] = pd.cut(
    df['age'], bins=[18, 30, 45, 60, 100], labels=labels, include_lowest=True
    )
df[['age', 'age_group']].head(10)
```

```python
# 1. Full range per group — resolves Senior boundary and confirms no overlap
df.groupby('age_group')['age'].agg(['min','max','count']).sort_index()

# 2. Full label inventory — confirms there are exactly 4 groups, no typos, no stray labels
df['age_group'].value_counts().sort_index()
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```
